```python
import numpy as np
# !nvidia-smi

import os
# os.environ["JAX_PLATFORMS"] = "cuda"
import jax
jax.config.update("jax_enable_x64", True)
import jax.numpy as jnp
import s2fft


# print("Available devices:", jax.devices())
# print("Default backend:", jax.default_backend())
# print("Jax version", jax.__version__, "64-bit", jax.config.read("jax_enable_x64"))

m = jnp.ones((17, 32))  # MWSS shape for L=16 is (L+1, 2L)
print("Original shape", m.shape)
flm = s2fft.transforms.spherical.forward(
    m, L=16, reality=True, method="jax", sampling="mwss"
)
print("L=16 works!", flm.shape)  # Works smoothly``
```
